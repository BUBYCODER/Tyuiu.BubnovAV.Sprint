## Название проекта

Вариант 9. Система заказов товаров и услуг


<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Продолжение проекта из Sprint1. Order хранит товары и считает сумму. OnlineOrder добавляет сведения о доставке, PhysicalOrder сообщает о возврате, SpecializedOrder учитывает скидку.

Добавлены по четыре свойства и метода к базовому классу и трём основным видам заказа. Простое наследование: Order → PhysicalOrder. Сложное: Order → OnlineOrder → ExpressOnlineOrder. OnlineOrder реализует два интерфейса: IPayable и ITrackable.

К версии Sprint2 добавлены ещё по три свойства и метода: подтверждение заказа, данные получателя, выдача из магазина и выполнение услуги. Вызовы через Order показывают полиморфизм. DisplayInfo переопределён, AddItem и CalculateTotal перегружены. OrderCollection<T> хранит заказы и позволяет добавить, удалить и вывести их.


#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте полиморфизм с перекрытием и прегегрузкой методов, а также generic классы

```
using System;
using System.Collections.Generic;
public static class Sprint3Task3
{
    public class Item
    {
        public string Name
        {
            get;
            set;
        }
        private decimal price;
        public decimal Price
        {
            get
            {
                return price;
            }
            set
            {
                if (value >= 0) price = value;
            }
        }
        public Item(string name, decimal price)
        {
            Name = name;
            Price = price;
        }
    }
    public interface IPayable
    {
        void Pay();
    }
    public interface ITrackable
    {
        void Track();
    }
    public class Order
    {
        public int Priority
        {
            get;
            set;
        }
        public DateTime DeliveryDate
        {
            get;
            set;
        }
        public bool IsConfirmed
        {
            get;
            private set;
        }
        public void SetPriority(int priority)
        {
            if (priority >= 1 && priority <= 3) Priority = priority;
        }
        public void SetDeliveryDate(DateTime date)
        {
            DeliveryDate = date;
        }
        public void Confirm()
        {
            IsConfirmed = true;
            Console.WriteLine("Заказ подтверждён");
        }
        public string CustomerName
        {
            get;
            set;
        }
        public string Status
        {
            get;
            set;
        }
        public string Comment
        {
            get;
            set;
        }
        public string Currency
        {
            get;
            set;
        }
        public void ChangeCustomer(string name)
        {
            CustomerName = name;
        }
        public void ChangeStatus(string status)
        {
            Status = status;
        }
        public void SetComment(string text)
        {
            Comment = text;
        }
        public virtual void DisplayInfo()
        {
            Console.WriteLine("Заказ " + OrderId + ", сумма: " + CalculateTotal());
        }
        public int OrderId
        {
            get;
            set;
        }
        public DateTime CreationDate
        {
            get;
            set;
        }
        public decimal TotalAmount
        {
            get;
            protected set;
        }
        private List<Item> items = new List<Item>();
        public Order(int id)
        {
            OrderId = id;
            CreationDate = DateTime.Today;
            Status = "Новый";
            Currency = "руб.";
        }
        public virtual decimal CalculateTotal()
        {
            TotalAmount = 0;
            foreach (Item item in items)
            TotalAmount += item.Price;
            return TotalAmount;
        }
        public virtual void AddItem(Item item)
        {
            items.Add(item);
            CalculateTotal();
            Console.WriteLine("Добавлен товар: " + item.Name);
        }
        public void AddItem(string name, decimal price)
        {
            AddItem(new Item(name, price));
        }
        public decimal CalculateTotal(decimal deliveryPrice)
        {
            return CalculateTotal() + deliveryPrice;
        }
        public virtual void RemoveItem(Item item)
        {
            if (items.Remove(item))
            {
                CalculateTotal();
                Console.WriteLine("Удалён товар: " + item.Name);
            }
            else
            Console.WriteLine("Такого товара нет в заказе");
        }
        protected bool HasItem(Item item)
        {
            return items.Contains(item);
        }
    }
    public class OnlineOrder : Order, IPayable, ITrackable
    {
        public override void DisplayInfo()
        {
            base.DisplayInfo();
            Console.WriteLine(CustomerEmail);
        }
        public string RecipientName
        {
            get;
            set;
        }
        public string ContactPhone
        {
            get;
            set;
        }
        public string NotificationText
        {
            get;
            set;
        }
        public void SetRecipient(string name)
        {
            RecipientName = name;
        }
        public void SetPhone(string phone)
        {
            ContactPhone = phone;
        }
        public void Notify(string text)
        {
            NotificationText = text;
            Console.WriteLine(CustomerEmail + ": " + NotificationText);
        }
        public string TrackingNumber
        {
            get;
            set;
        }
        public string PaymentMethod
        {
            get;
            set;
        }
        public string Website
        {
            get;
            set;
        }
        public bool IsPaid
        {
            get;
            private set;
        }
        public void Pay()
        {
            IsPaid = true;
            Status = "Оплачен";
            Console.WriteLine("Заказ оплачен: " + PaymentMethod);
        }
        public void Track()
        {
            Console.WriteLine("Номер доставки: " + TrackingNumber);
        }
        public void ChangeEmail(string email)
        {
            CustomerEmail = email;
        }
        public void ShowDelivery()
        {
            Console.WriteLine("Доставка: " + DeliveryMethod);
        }
        public string CustomerEmail
        {
            get;
            set;
        }
        public string DeliveryMethod
        {
            get;
            set;
        }
        public OnlineOrder(int id, string email, string delivery) : base(id)
        {
            CustomerEmail = email;
            DeliveryMethod = delivery;
        }
        public override void AddItem(Item item)
        {
            base.AddItem(item);
            Console.WriteLine("Доставка: " + DeliveryMethod);
            Console.WriteLine("Email: " + CustomerEmail);
        }
        public void ChangeDelivery(string method)
        {
            DeliveryMethod = method;
        }
    }
    public class PhysicalOrder : Order
    {
        public override void DisplayInfo()
        {
            base.DisplayInfo();
            Console.WriteLine(DeliveryAddress);
        }
        public int ShelfNumber
        {
            get;
            set;
        }
        public bool IsPacked
        {
            get;
            private set;
        }
        public bool IsCollected
        {
            get;
            private set;
        }
        public void SetShelf(int number)
        {
            if (number > 0) ShelfNumber = number;
        }
        public void Pack()
        {
            IsPacked = true;
            Console.WriteLine("Заказ упакован");
        }
        public void Collect()
        {
            if (IsPacked)
            {
                IsCollected = true;
                Console.WriteLine("Заказ выдан");
            }
        }
        public string StoreName
        {
            get;
            set;
        }
        public string SellerName
        {
            get;
            set;
        }
        public string ReceiptNumber
        {
            get;
            set;
        }
        public DateTime PickupDate
        {
            get;
            set;
        }
        public void ChangeStore(string name)
        {
            StoreName = name;
        }
        public void ChangeSeller(string name)
        {
            SellerName = name;
        }
        public void SetPickupDate(DateTime date)
        {
            PickupDate = date;
        }
        public void PrintReceipt()
        {
            Console.WriteLine("Чек " + ReceiptNumber + ": " + CalculateTotal());
        }
        public string DeliveryAddress
        {
            get;
            set;
        }
        public PhysicalOrder(int id, string address) : base(id)
        {
            DeliveryAddress = address;
        }
        public override void RemoveItem(Item item)
        {
            bool found = HasItem(item);
            base.RemoveItem(item);
            if (found)
            Console.WriteLine("Возврат товара. Адрес доставки: " + DeliveryAddress);
        }
        public void ChangeAddress(string address)
        {
            DeliveryAddress = address;
        }
    }
    public class SpecializedOrder : Order
    {
        public override void DisplayInfo()
        {
            base.DisplayInfo();
            Console.WriteLine(SpecialConditions);
        }
        public string ServiceName
        {
            get;
            set;
        }
        public string CustomerRequest
        {
            get;
            set;
        }
        public bool IsCompleted
        {
            get;
            private set;
        }
        public void SetService(string name)
        {
            ServiceName = name;
        }
        public void SetRequest(string request)
        {
            CustomerRequest = request;
        }
        public void Complete()
        {
            IsCompleted = true;
            Console.WriteLine("Услуга выполнена: " + ServiceName);
        }
        public string Category
        {
            get;
            set;
        }
        public string ManagerName
        {
            get;
            set;
        }
        public DateTime Deadline
        {
            get;
            set;
        }
        public bool IsApproved
        {
            get;
            private set;
        }
        public void ChangeCategory(string category)
        {
            Category = category;
        }
        public void AssignManager(string name)
        {
            ManagerName = name;
        }
        public void SetDeadline(DateTime date)
        {
            Deadline = date;
        }
        public void Approve()
        {
            IsApproved = true;
            Console.WriteLine("Заказ согласован");
        }
        public string SpecialConditions
        {
            get;
            private set;
        }
        private decimal discount;
        public SpecializedOrder(int id, decimal discount) : base(id)
        {
            SetDiscount(discount);
        }
        public void SetDiscount(decimal value)
        {
            if (value >= 0 && value <= 100)
            {
                discount = value;
                SpecialConditions = "Скидка " + value + "%";
                CalculateTotal();
            }
        }
        public override decimal CalculateTotal()
        {
            decimal sum = base.CalculateTotal();
            TotalAmount = sum - sum * discount / 100;
            return TotalAmount;
        }
    }
    public class Customer
    {
        public string Name
        {
            get;
            set;
        }
        public Customer(string name)
        {
            Name = name;
        }
        public void Buy(Order order, Item item)
        {
            Console.WriteLine(Name + " покупает " + item.Name);
            order.AddItem(item);
        }
    }
    public class ExpressOnlineOrder : OnlineOrder
    {
        public int DeliveryHours
        {
            get;
            set;
        }
        public string CourierName
        {
            get;
            set;
        }
        public ExpressOnlineOrder(int id, string email, string delivery, int hours) : base(id, email, delivery)
        {
            DeliveryHours = hours;
        }
        public override void DisplayInfo()
        {
            base.DisplayInfo();
            Console.WriteLine("Экспресс-доставка за " + DeliveryHours + " ч.");
        }
    }
    public class OrderCollection<T> where T : Order
    {
        private List<T> orders = new List<T>();
        public void Add(T order)
        {
            orders.Add(order);
        }
        public void Remove(T order)
        {
            orders.Remove(order);
        }
        public void DisplayOrders()
        {
            foreach (T order in orders) order.DisplayInfo();
        }
    }
    public static void Run()
    {
        Item book = new Item("Книга", 500);
        Item pen = new Item("Ручка", 100);
        OnlineOrder online = new OnlineOrder(1, "student@example.com", "Курьер");
        PhysicalOrder physical = new PhysicalOrder(2, "Тюмень, ул. Республики, 1");
        SpecializedOrder special = new SpecializedOrder(3, 10);
        Order[] orders =
        {
            online, physical, special
        }
        ;
        foreach (Order order in orders)
        {
            order.AddItem(book);
            order.AddItem(pen);
            Console.WriteLine("Заказ " + order.OrderId + ": " + order.CalculateTotal());
        }
        physical.RemoveItem(pen);
        Console.WriteLine("После возврата: " + physical.CalculateTotal());
        Console.WriteLine(special.SpecialConditions);
        Customer customer = new Customer("Андрей");
        Item pencil = new Item("Карандаш", 50);
        customer.Buy(online, pencil);
        Console.WriteLine("Сумма онлайн-заказа: " + online.CalculateTotal());
        online.ChangeCustomer("Андрей");
        online.ChangeStatus("Оформлен");
        online.SetComment("Доставить после занятий");
        online.PaymentMethod = "Карта";
        online.TrackingNumber = "12345";
        online.Website = "Магазин книг";
        online.ChangeEmail("andrey@example.com");
        online.ShowDelivery();
        IPayable payment = online;
        payment.Pay();
        ITrackable tracking = online;
        tracking.Track();
        physical.ChangeStore("Книги");
        physical.ChangeSeller("Иван");
        physical.SetPickupDate(DateTime.Today.AddDays(1));
        physical.ReceiptNumber = "15";
        physical.PrintReceipt();
        special.ChangeCategory("Учебные товары");
        special.AssignManager("Ольга");
        special.SetDeadline(DateTime.Today.AddDays(7));
        special.Approve();
        ExpressOnlineOrder express = new ExpressOnlineOrder(4, "andrey@example.com", "Курьер", 2);
        express.AddItem(book);
        express.DisplayInfo();
        online.SetPriority(2);
        online.SetDeliveryDate(DateTime.Today.AddDays(2));
        online.Confirm();
        online.SetRecipient("Андрей");
        online.SetPhone("Не указан");
        online.Notify("Заказ готов к доставке");
        physical.SetShelf(5);
        physical.Pack();
        physical.Collect();
        special.SetService("Подбор учебных материалов");
        special.SetRequest("По программе третьего семестра");
        special.Complete();
        online.AddItem("Тетрадь", 50);
        Console.WriteLine("Сумма с доставкой: " + special.CalculateTotal(100));
        OrderCollection<Order> collection = new OrderCollection<Order>();
        foreach (Order order in orders) collection.Add(order);
        collection.DisplayOrders();
        collection.Remove(physical);
        Console.WriteLine("После удаления физического заказа:");
        collection.DisplayOrders();
    }
}
Sprint3Task3.Run();
```
